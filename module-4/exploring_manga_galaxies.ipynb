## An aside on *FITS* for the astro-interested

*Feel free to ignore this*

We've introduced `pandas` as a powerful way to ingest and manipulate large data sets, but astronomers have their own preferences for doing this.  Astronomers typically store and transfer data in the form of "[FITS files](https://fits.gsfc.nasa.gov/fits_primer.html)." Data contained within fits files are stored as either multidimensional **images** or as data tables, called **binary tables**. A single fits file can contain both imaging data and binary tables arranged in separate **extensions** of the file. 

The FITS format improves on CSV by including columns names and data types explicitly (even units, I believe).  It also allows for headers and keywords which are useful for storing metadata (i.e., background information about the data itself, how it was taken, processed, etc.).  Finally, it can handle multi-dimensional data sets.

(In the non-astro world, these capabilities can be achieved with `xarray` which is an extension to `pandas`.)

Tools for handling FITS files are including in the `astropy` package, when you can install with `> conda install astropy`

```python
# OPTIONAL CELL on FITS Files: Here's how to ingest FITS files
from astropy.io import fits
from astropy.table import Table

# To generically open all HDUs of a FITS file
hdu = fits.open('drpall-v3_1_1.fits')
# hdu.info()  # Tell us what's inside
# hdu[1].data  # Access the data in a particular HDU

# To pull out a binary table from specific HDU
#   The dat object is analogous to a dataframe (but the methods and properties are different)
dat = Table.read('drpall-v3_1_1.fits', format='fits', hdu=1)
```

## A color-mass diagram of observed MaNGA galaxies

A very common diagnostic of a population of galaxies is a comparison of the flux each galaxy has in different wavelength ranges, or **bands**.  Galaxies with lots of young stars are "bluer." Their flux at shorter wavelengths is greater than their flux at redder wavelengths.  Galaxies dominated by old stars are "redder."  They emit more flux at red wavelengths than at blue wavelengths.  A simple color measurement, then, can tell us roughly how old a galaxy is and whether or not it's actively making new, young stars.

### An (optional) aside on magnitudes for the astro-curious...

For historical reasons, astronomers typically refer to the flux of an object in terms of its **magnitude**.  The magnitude is related to an inverted **logarithm of the object's flux**.  An object has both an **apparent magnitude** -- the brightness of the object as we see it from the Earth -- and an **absolute magnitude** -- the brightness of an object from a fixed distance.  As you might have guessed, the most physically interesting quantity is the latter.  I.e., it's typically more physically meaningful to know how bright an object is relative to others in an absolute sense as opposed to how bright it is relative to other objects as viewed from the Earth.

For historical reasons, the magnitude system is on an inverted logarithmic scale.  A logarithmic scale is one that is incremented by the product of the scale's base, instead of the sum.  For example, the following is a set that increments linearly:

$$x = \{ 2, 4, 6, 8, 10, ... \}; $$

each element is the previous element with 2 added to it.  In contrast, the following is a set that increments logarithmically:

$$y = \{ 2, 4, 8, 16, 32, ...\};$$

each element is the previous element *multiplied* by two.  If you take the base-2 logarithm of the set $y$ you get:

$$Y = log_2(y) = \{ 1, 2, 3, 4, 5, ...\}$$

So a logarithmic scale is just a scale that increases linearly in the logarithm.

For the magnitude system, it's inverted in the sense that **brighter objects have smaller magnitude**.  In the example above, you can think of this as creating a set where each element is the previous element *divided* by the logarithm's base.  It's also import to know that, because magnitudes are based on the logarithm of the flux, the difference between two magnitudes provides you with the logarithm of the ratio of two fluxes.  In the example above, if you take the difference between the first and second elements in set $Y$ you find, $Y_1 - Y_0 = 1$, which is the same thing as taking the logarithm of the ratio between the first two elements of set $y$: $log_2(y_1/y_0) = 1$.

### For the astro-curious: the "red sequence" and the "blue cloud"

You may notice that the data points in the color-magnitude or color-mass diagram lie in roughly two groups: For the color-magnitude diagram, we see a sequence of galaxies with $-18.5 > M_i > -23.5$ and $N-r \sim 5.5$ and a more distributed "cloud" of galaxies toward smaller $N-r$.  

Astronomers call these two features the "red sequence" and the "blue cloud", respectively.  The red sequence galaxies are a population that (for reasons we still don't fully understand!) has stopped forming stars (hence, all the remaining stars are old and red).  The blue cloud galaxies are still activately turning gas into new stars.  They have a mix of young and old stars, and that makes them bluer.

You can see from the density of points that low-mass galaxies tend to be in the "blue cloud" while high-mass galaxies tend to be on the "red sequence".  This is a key result.  Some how, the processes that kill off star formation are happening more in the massive galaxies.  Super-massive black holes?  Radio feedback?  Galaxy mergers?  Galaxy environments?  These are all active questions in this field right now.
